```python
# Simulador de IA: Asistente de Evaluación para Préstamo Personal v1.1
# Nivel: Principiante (Sin funciones, sin objetos, sin ciclos for, sin try)

print("====================================================")
print("   BIENVENIDO AL SISTEMA DE IA 'PRESTA-BOT' 2026   ")
print("====================================================")
print("Nota: Este sistema evalúa el riesgo crediticio de forma automática.")
print("----------------------------------------------------")

ejecutando_ia = "si"

# CORRECCIÓN: Se cambiaron las "o" por "or" que es el estándar de Python
while ejecutando_ia == "si" or ejecutando_ia == "sí" or ejecutando_ia == "SI":
    print("\n[IA]: Iniciando evaluación de nuevo cliente...")

    # --- FASE 1: Recopilación de Datos (Sensores de la IA) ---
    print("Por favor, ingresa los datos solicitados en NÚMEROS ENTEROS.")

    edad = int(input("- Ingresa la edad del solicitante: "))
    ingresos = int(input("- Ingresa los ingresos mensuales del cliente (en tu moneda): "))
    historial_negativo = input("- ¿El cliente tiene deudas vencidas o mal historial? (si/no): ")
    monto_solicitado = int(input("- ¿Qué monto de dinero está solicitando?: "))

    print("\n[IA]: Procesando datos financieros y calculando riesgo...")

    # --- FASE 2: Motor de Decisiones (Lógica con IF / ELIF / ELSE) ---

    # Regla de rechazo automático 1: Menor de edad o adulto mayor avanzado
    # CORRECCIÓN: Se cambió "o" por "or"
    if edad < 18 or edad > 75:
        print("\n--- CRITERIO RECHAZADO ---")
        print("Resultado del análisis: El solicitante no cumple con la edad requerida (18-75 años).")
        print("Decisión de la IA: PRÉSTAMO DENEGADO.")

    # Regla de rechazo automático 2: Mal historial crediticio (Alto Riesgo)
    elif historial_negativo == "si":
        print("\n--- CRITERIO RECHAZADO ---")
        print("Resultado del análisis: El historial crediticio presenta alertas rojas o deudas activas.")
        print("Decisión de la IA: PRÉSTAMO DENEGADO por riesgo de impago.")

    # Regla de aprobación 1: Cliente Premium (Altos ingresos y pide algo razonable)
    # CORRECCIÓN: Se usó "and" para evaluar ambas condiciones juntas
    elif ingresos >= 3000 and monto_solicitado <= (ingresos * 5):
        print("\n--- CRITERIO APROBADO: CLIENTE PREMIUM ---")
        print("Resultado del análisis: Capacidad de pago excelente y perfil de muy bajo riesgo.")
        print("Decisión de la IA: PRÉSTAMO APROBADO INMEDIATAMENTE.")
        print("Recomendación: Ofrecer la tasa de interés más baja disponible.")

    # Regla de aprobación 2: Cliente Estándar (Ingresos medios y monto moderado)
    elif ingresos >= 1000 and monto_solicitado <= (ingresos * 3):
        print("\n--- CRITERIO APROBADO: CLIENTE ESTÁNDAR ---")
        print("Resultado del análisis: El monto solicitado es adecuado para el nivel de ingresos.")
        print("Decisión de la IA: PRÉSTAMO APROBADO.")
        print("Recomendación: Aplicar tasa de interés estándar con seguro de desempleo.")

    # Regla de revisión manual: Ingresos bajos o pide demasiado dinero
    elif monto_solicitado > (ingresos * 5):
        print("\n--- CRITERIO EN REVISIÓN ---")
        print("Resultado del análisis: El monto solicitado supera 5 veces los ingresos del cliente.")
        print("Decisión de la IA: PRÉSTAMO DETENIDO.")
        print("Recomendación: Requiere un aval (garantía) o reducir el monto solicitado.")

    # Cualquier otro caso que no encaje en las reglas anteriores
    else:
        print("\n--- CRITERIO RECHAZADO ---")
        print("Resultado del análisis: Los ingresos son insuficientes para el monto solicitado.")
        print("Decisión de la IA: PRÉSTAMO DENEGADO.")

    print("----------------------------------------------------")

    # --- FASE 3: Control de interactividad ---
    ejecutando_ia = input("\n¿Deseas evaluar a otro cliente? (si/no): ")

print("\n====================================================")
print("  SISTEMA APAGADO. ¡Gracias por usar Presta-Bot!    ")
print("====================================================")
```
